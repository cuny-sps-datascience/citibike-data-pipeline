```python
# The purpose of this notebook is to try to improve the cost efficiency of the frequent prediction task that I have been
# running. Basically, I want to see if there is a big fixed cost in starting up a serverless workspace that can be reduced
# by chaining together the three notebooks in that pipeline. It also may be possible to increase the speed by having one
# serverless workspace perform all three tasks.


import time

PIPELINE_PARAMETERS = ["run_mode", "schema", "status_table", "info_table", "publish_page"]

# Widget parameters
step_arguments = {}
for parameter_name in PIPELINE_PARAMETERS:
    dbutils.widgets.text(parameter_name, "")
    if dbutils.widgets.get(parameter_name):
        step_arguments[parameter_name] = dbutils.widgets.get(parameter_name)
print("Arguments:", step_arguments)

# The notebooks we will execute
PIPELINE_STEPS = [
    "./1-citibike_json_status_processor",
    "./2-citibike-station-status-silver",
    "./free_edition_models/5-predict-stations-FE",
]

for notebook_path in PIPELINE_STEPS:
    step_start = time.time()
    dbutils.notebook.run(notebook_path, timeout_seconds=900, arguments=step_arguments)
    print(f"{notebook_path}: {time.time() - step_start:.0f} s")
```
